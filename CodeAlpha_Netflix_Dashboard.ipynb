```python
# ==========================================
# NETFLIX MOVIES & TV SHOWS DASHBOARD
# DATA VISUALIZATION USING PYTHON
# ==========================================

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
plt.rcParams["figure.figsize"] = (10,6)

# ==========================================
# LOAD DATASET
# ==========================================

# Change the path if your file is inside a folder
df = pd.read_csv('/content/drive/MyDrive/netflix_titles.csv')

# ==========================================
# DATA OVERVIEW
# ==========================================

print("="*60)
print("DATASET SHAPE")
print(df.shape)

print("\nCOLUMNS")
print(df.columns)

print("\nDATA TYPES")
print(df.dtypes)

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nFIRST FIVE RECORDS")
print(df.head())

# ==========================================
# DATA CLEANING
# ==========================================

df['country'] = df['country'].fillna("Unknown")
df['rating'] = df['rating'].fillna("Unknown")
df['director'] = df['director'].fillna("Unknown")
df['cast'] = df['cast'].fillna("Unknown")

# ==========================================
# DASHBOARD
# ==========================================

fig, axes = plt.subplots(3,2,figsize=(18,18))

# ==========================================
# 1. MOVIES VS TV SHOWS
# ==========================================

sns.countplot(
    data=df,
    x='type',
    palette='Set2',
    ax=axes[0,0]
)

axes[0,0].set_title("Movies vs TV Shows")
axes[0,0].set_xlabel("Content Type")
axes[0,0].set_ylabel("Count")

# ==========================================
# 2. CONTENT BY COUNTRY
# ==========================================

country = (
    df['country']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

sns.barplot(
    x=country.values,
    y=country.index,
    palette='viridis',
    ax=axes[0,1]
)

axes[0,1].set_title("Top 10 Countries")
axes[0,1].set_xlabel("Number of Titles")

# ==========================================
# 3. RATING DISTRIBUTION
# ==========================================

rating = df['rating'].value_counts().head(10)

sns.barplot(
    x=rating.index,
    y=rating.values,
    palette='coolwarm',
    ax=axes[1,0]
)

axes[1,0].set_title("Rating Distribution")
axes[1,0].tick_params(axis='x', rotation=45)

# ==========================================
# 4. TOP GENRES
# ==========================================

genre = (
    df['listed_in']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

sns.barplot(
    x=genre.values,
    y=genre.index,
    palette='magma',
    ax=axes[1,1]
)

axes[1,1].set_title("Top 10 Genres")
axes[1,1].set_xlabel("Count")

# ==========================================
# 5. RELEASES BY YEAR
# ==========================================

year = df['release_year'].value_counts().sort_index()

sns.lineplot(
    x=year.index,
    y=year.values,
    marker='o',
    ax=axes[2,0]
)

axes[2,0].set_title("Releases by Year")
axes[2,0].set_xlabel("Release Year")
axes[2,0].set_ylabel("Number of Titles")

# ==========================================
# REMOVE EMPTY PLOT
# ==========================================

axes[2,1].axis('off')

plt.suptitle(
    "Netflix Movies & TV Shows Dashboard",
    fontsize=20,
    fontweight="bold"
)

plt.tight_layout()

plt.show()

# ==========================================
# SUMMARY
# ==========================================

print("="*60)
print("SUMMARY")
print("="*60)

print("Total Titles :", len(df))
print("Movies :", (df['type']=="Movie").sum())
print("TV Shows :", (df['type']=="TV Show").sum())

print("\nTop Country")
print(country.head())

print("\nTop Genres")
print(genre.head())

print("\nMost Common Ratings")
print(rating.head())

print("\nLatest Release Year :", df['release_year'].max())
print("Oldest Release Year :", df['release_year'].min())

print("="*60)
print("DASHBOARD CREATED SUCCESSFULLY")
print("="*60)
```
